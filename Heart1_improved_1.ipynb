# Data Preprocessing

### Outlier Handling

```python
# ── Outlier Handling (IQR Winsorization) ──
# Applied to all continuous numerical features to reduce skew

def whisker(col):
    Q1 = col.quantile(0.25)
    Q3 = col.quantile(0.75)
    IQR = Q3 - Q1
    lb = Q1 - 1.5 * IQR
    ub = Q3 + 1.5 * IQR
    return lb, ub

# Apply Winsorization to all continuous features
continuous_cols = ['BMI', 'PhysicalHealth', 'MentalHealth', 'SleepTime']

for col in continuous_cols:
    lb, ub = whisker(df[col])
    df[col] = np.where(df[col] < lb, lb, df[col])
    df[col] = np.where(df[col] > ub, ub, df[col])
    print(f'✅ {col}: capped to [{lb:.2f}, {ub:.2f}]')
```

### Encoding

```python
# Identify categorical columns in the dataset
categorical_columns = df.select_dtypes(include=['object']).columns
print(f"\n                  ----Categorical Columns----                  \n\n{categorical_columns}")
print()

for i in categorical_columns:
    unique = df[i].unique()
    a = len(unique)
    print(f"---- {i} ----\n")
    print(f"The column has a total of : {a} unique values")
    print(f"{unique}\n")
```

```python
df = df.rename(
    columns={
        "Smoking" : "IS_Smoking",
        "AlcoholDrinking" : "Alcoholic",
        "Stroke" : "IS_Stroke",
        "DiffWalking" : "IS_DiffWalking",
        "Sex" : "IS_MALE",
        "PhysicalActivity" : "IS_PhysicalActivity",
        "Asthma" : "IS_Asthma",
        "KidneyDisease" : "IS_KidneyDisease",
        "SkinCancer" : "IS_SkinCancer",
        'Diabetic' : "IS_Diabetic"
    }
)
df.columns
```

```python
# Applying Encoding

# One-hot encoding
df = pd.get_dummies(df, columns=["Race"],dtype="int")

# Label encoding
df["IS_MALE"] = df["IS_MALE"].map({
    "Male": 1,
    "Female": 0
})

df["IS_Smoking"] = df["IS_Smoking"].map({
    "Yes": 1,
    "No": 0
})

df["IS_Stroke"] = df["IS_Stroke"].map({
    "Yes": 1,
    "No": 0
})

df["IS_DiffWalking"] = df["IS_DiffWalking"].map({
    "Yes": 1,
    "No": 0
})

df["IS_PhysicalActivity"] = df["IS_PhysicalActivity"].map({
    "Yes": 1,
    "No": 0
})

df["IS_Asthma"] = df["IS_Asthma"].map({
    "Yes": 1,
    "No": 0
})

df["IS_KidneyDisease"] = df["IS_KidneyDisease"].map({
    "Yes": 1,
    "No": 0
})

df["IS_SkinCancer"] = df["IS_SkinCancer"].map({
    "Yes": 1,
    "No": 0
})

df["HeartDisease"] = df["HeartDisease"].map({
    "Yes": 1,
    "No": 0
})

df["Alcoholic"] = df["Alcoholic"].map({
    "Yes": 1,
    "No": 0
})

df["GenHealth"] = df["GenHealth"].map({
    "Poor": 0,
    "Fair": 1,
    "Good": 2,
    "Very good": 3,
    "Excellent": 4
})

# Ordinal encoding: captures gradual risk increase
df["IS_Diabetic"] = df["IS_Diabetic"].map({
    "No": 0,
    "No, borderline diabetes": 1,
    "Yes (during pregnancy)": 2,
    "Yes": 3
})

df["AgeCategory"] = df["AgeCategory"].map({
    "18-24": 0,
    "25-29": 1,
    "30-34": 2,
    "35-39": 3,
    "40-44": 4,
    "45-49": 5,
    "50-54": 6,
    "55-59": 7,
    "60-64": 8,
    "65-69": 9,
    "70-74": 10,
    "75-79": 11,
    "80 or older": 12
})

df.dtypes
```

### Feature Engineering

```python
# ── Feature Engineering ──

# Overall health burden (physical + mental)
df["HealthScore"] = df["PhysicalHealth"] + df["MentalHealth"]

# Lifestyle risk (smoking + alcohol + no activity)
df["LifestyleRisk"] = df["IS_Smoking"] + df["Alcoholic"] + (1 - df["IS_PhysicalActivity"])

# Total number of co-morbidities
df["DiseaseCount"] = df["IS_Stroke"] + df["IS_Diabetic"] + df["IS_Asthma"] + df["IS_KidneyDisease"] + df["IS_SkinCancer"]

# Obesity indicator
df["BMIRisk"] = (df["BMI"] > 30).astype(int)

# Poor sleep indicator
df["SleepRisk"] = ((df["SleepTime"] < 5) | (df["SleepTime"] > 9)).astype(int)

# Age and disease interaction
df["Age_Disease"] = df["AgeCategory"] * df["DiseaseCount"]

# Final total risk score
# NOTE: MobilityRisk removed — it was a duplicate of IS_DiffWalking
df["TotalRisk"] = df["LifestyleRisk"] + df["DiseaseCount"] + df["BMIRisk"] + df["SleepRisk"] + df["IS_DiffWalking"]
```

### Train Test Split

```python
#Splitting
from sklearn.model_selection import train_test_split
x_train , x_test , y_train , y_test = train_test_split( X , y , test_size=0.3 , random_state = 42 )# Model Training
print(f'Train size: {x_train.shape[0]:,}  |  Test size: {x_test.shape[0]:,}')
```

# Models Training

```python
# ===============================
# 🤖 MODEL SELECTION + TRAINING
# ===============================

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from lightgbm import LGBMClassifier
from xgboost import XGBClassifier

# ------------------------------
# 0. Class Imbalance Handling
# ------------------------------

# Compute imbalance ratio for scale_pos_weight
neg_count = (y_train == 0).sum()
pos_count = (y_train == 1).sum()
imbalance_ratio = neg_count / pos_count
print(f'Imbalance ratio (neg/pos): {imbalance_ratio:.2f}')

# ------------------------------
# 1. Select Models
# ------------------------------
models = {
    "Logistic Regression": LogisticRegression(class_weight='balanced', max_iter=1000),
    "Decision Tree": DecisionTreeClassifier(max_depth=15, class_weight='balanced'),
    "Random Forest": RandomForestClassifier(class_weight='balanced'),
    "LightGBM": LGBMClassifier(
        n_estimators=300,
        learning_rate=0.05,
        scale_pos_weight=imbalance_ratio,  # native imbalance handling
    ),
        "XGBoost": XGBClassifier(
        n_estimators=300,
        learning_rate=0.05,
        max_depth=6,
        scale_pos_weight=imbalance_ratio,
        random_state=42,
        eval_metric='logloss',
        use_label_encoder=False
    )
}

# ------------------------------
# 2. Train Models
# ------------------------------
trained_models = {}

for name, model in models.items():
    model.fit(x_train, y_train)
    trained_models[name] = model

print("✅ All models trained successfully!")
```

# Models Evalution

```python
from sklearn.metrics import accuracy_score, f1_score, classification_report, confusion_matrix

results = {}

for name, model in models.items():

    # Prediction
    y_pred = model.predict(x_test)

    # Metrics
    acc = accuracy_score(y_test, y_pred)
    f1 = f1_score(y_test, y_pred, average='weighted')

    results[name] = {"Accuracy": acc, "F1 Score": f1}

    # Detailed Evaluation
    print(f"\n===== {name} =====")
    print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred))
    print("\nClassification Report:\n", classification_report(y_test, y_pred))

# Final Comparison
print("\n=== FINAL RESULTS ===")
for model, scores in results.items():
    print(f"{model}: {scores}")
```

# Hyperparameter Tuning

```python
from sklearn.model_selection import GridSearchCV

# Base model — CHANGE: scale_pos_weight instead of class_weight for consistency
lgbm = LGBMClassifier(
    scale_pos_weight=imbalance_ratio,
    random_state=42
)

# Parameter grid — PRESERVED from original
params = {
    "num_leaves":    [31, 50, 70],
    "max_depth":     [10, 20, -1],
    "learning_rate": [0.01, 0.05, 0.1],
    "n_estimators":  [100, 200]
}

grid = GridSearchCV(
    estimator=lgbm,
    param_grid=params,
    scoring='recall',   # PRESERVED — optimizing for sensitivity in medical context
    cv=3,
    n_jobs=-1,
    verbose=1           # ADDED: show progress
)

grid.fit(x_train, y_train)
best_model = grid.best_estimator_

print("Best Parameters:", grid.best_params_)
```

```python
y_prob = best_model.predict_proba(x_test)[:,1]

y_pred = (y_prob > 0.53).astype(int)

print(f"\n===== {"LightGBM"} =====")
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))
```
